```python

import kagglehub
import os
from pathlib import Path
```

```python
# Download the Asian vs African Elephant dataset
path = kagglehub.dataset_download( "vivmankar/asian-vs-african-elephant-image-classification")
print("Dataset downloaded at:", path)
```

```python
# Inspect the downloaded folder structure
dataset_path = Path(path)
dataset_path
```

```python
# Inspect folders and count images
for root, dirs, files in os.walk(path):
    image_files = [file for file in files 
                 if file.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".webp"))]

    
    print(f"\nFolder: {os.path.basename(root)}")
    print(f"Image count: {len(image_files)}")
    print(f"Subfolders: {dirs}")
```

```python
# Dataset Folder Structure
#
# 1/                         → KaggleHub downloaded dataset version
# └── dataset/               → Main dataset directory
#     ├── test/              → Test dataset
#     │   ├── African/       → 97 African elephant images
#     │   └── Asian/         → 91 Asian elephant images
#     │
#     └── train/             → Training dataset
#         ├── African/       → 420 African elephant images
#         └── Asian/         → 420 Asian elephant images
```
